**_Accessing Elements of Lists_**

Use a pair of brackets ([ ]) to access elements from a list. Lists in Python are zero-indexed, meaning the first element starts at position 0.

To extract a subset of elements in a list use **_slice_** notation. Slice uses the following syntax:

list[start:stop]

An important syntax note is that the **_start_** index is inclusive, but the **_stop_** index is exclusive. In other words, [0:2] returns the first and second element in a list, but **_not_** the third.

We can leave either side empty in slice notation, which will either give us every item in a list up until some element number.

list[:stop]

Or it will give us every item in a list from some element number through the rest of the list.

list[start:]

Let's start with a simple example where we loop over numbers in a list and print out each number.

```python
#Simple for loop example
numbers = [1, 2, 3]
for number in numbers:
    print(number)

#'long' way of creating the output (p. 95 in text)
#delete this and insert your code hree
```

**_Fun with loops_**

The **len()** function can be used to return the number of rows in a data frame.

To loop over the rows of a **pandas** data frame, the Python function that is important to know is **iloc[ ]**. **iloc[ ]** allows you to select rows and columns of a data frame using specific integer positions. **iloc[ ]** stands for integer location.

For all indexing methods, we provide the row indexer first and then the column indexer, with a comma separating them:
df.loc[row_indexer, column_indexer]. **iloc[ ]** is exclusive of endpoints, meaning it will not include the number the specify, rather, it will include the number right before the number you specify.

**iloc[ ]** works on the positions in your index. This means that if you give in iloc[2], it will look for the values of your data frame that are at index 2.

If you use a : (colon) as one of the arguments it means you want to return all rows or all columns.

```python
#reminder how the df_students data frame looks like
df_students.head()
#use the len() function to find the total number of rows in the df_students data frame
len(df_students)

#use iloc[] to extract the value of the first row of the second column.
df_students.iloc[0,1]

#use iloc[] to extract the value of the third row of the first column
df_students.iloc[2,0]

#use iloc to extract all columns of the fourth row (data for the fourth row)
df_students.iloc[3,:]

#use iloc to extract all values of the fourth column
df_students.iloc[:,3]

#use the range, length, and iloc[] functions to write a for loop that loops over the rows of the df_students data frame and adds 2 to the Calculus scores
for i in range(len(df_students)):
  new_calc=df_students.iloc[1,5]+2
  print(1)
  print(new_calc)
```
